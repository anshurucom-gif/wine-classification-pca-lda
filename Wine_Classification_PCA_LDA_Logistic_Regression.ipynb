```python
#apply PCA with 10 features
pca_95= PCA(n_components=10)

x_train_pca= pca_95.fit_transform(X_train_std)
x_test_pca= pca_95.transform(X_test_std)
```

```python
print("Original training shape:", X_train_std.shape)
print("PCA training shape:", x_train_pca.shape)

print("Original testing shape:", X_test_std.shape)
print("PCA testing shape:", x_test_pca.shape)
```

```python
#logistic regression using PCA
log_pca= LogisticRegression(max_iter=1000, random_state=42)

log_pca.fit(x_train_pca, Y_train)
```

```python
y_pred_pca= log_pca.predict(x_test_pca)
```

PCA is able to help us produce same result by using 3 less features

```python
#LDA

from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
lda= LinearDiscriminantAnalysis(n_components=2)
```

```python
x_train_LDA= lda.fit_transform(X_train_std, Y_train)
```

```python
x_test_LDA= lda.transform(X_test_std)
```

```python
log_lda = LogisticRegression(max_iter=1000, random_state=42)

log_lda.fit(x_train_LDA, Y_train)

y_pred_lda = log_lda.predict(x_test_LDA)
```

```python
LDA_accuracy= accuracy_score(Y_test, y_pred_lda)
```

```python
print("LDA Model Accuracy:", accuracy_score(y_test, y_pred_lda))

print("\nClassification Report:")
print(classification_report(y_test, y_pred_lda))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_lda))
```
